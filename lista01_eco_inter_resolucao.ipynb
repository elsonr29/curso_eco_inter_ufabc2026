**1.a)** Elabore a função de produção para cada país (local e estrangeiro). Em seguida, calcule a produção máxima de ambos os bens para os países em autarquia e para a economia mundial. Por fim, represente a fronteira de possibilidades de produção.

Consideramos que o país local e o estrangeiro possuem funções de produção próprias. Não há comércio. A função de produção origina a alocação de trabalho ($L$) entre os setores de manufaturas e agricultura ($L_m$ e $L_a$), dada a necessidade unitária de trabalho ($a_m$ e $a_a$). Podemos escrever as relações como:

**País Local:**

Função de produção:

$$Q_m = \frac{L_m}{a_m} \quad \text{e} \quad Q_a = \frac{L_a}{a_a}$$

A restrição de alocação do fator trabalho ($L = L_m + L_a$) determina a fronteira de possibilidades de produção:

$$a_m Q_m + a_a Q_a \le L$$

Substituindo os valores ($L = 1200$, $a_m = 3$ e $a_a = 2$):

$$Q_m = \frac{L_m}{3} \quad \text{e} \quad Q_a = \frac{L_a}{2}, \quad \text{tal que } L_m + L_a \le 1200$$

Produção máxima de manufaturas ($L_m = 1200$ e $L_a = 0$):

$$Q_m = \frac{1200}{3} \quad \text{e} \quad Q_a = \frac{0}{2}, \quad \text{tal que } L_m + L_a \le 1200$$

$$Q_m = 400 \quad \text{e} \quad Q_a = 0$$

Produção máxima de agricultura($L_m = 0$ e $L_a = 1200$):

$$Q_m = \frac{0}{3} \quad \text{e} \quad Q_a = \frac{1200}{2}, \quad \text{tal que } L_m + L_a \le 1200$$

$$Q_m = 0 \quad \text{e} \quad Q_a = 600$$


**País Estrangeiro:**

As funções de produção setoriais são:

$$Q_m^* = \frac{L_m^*}{a_m^*} \quad \text{e} \quad Q_a^* = \frac{L_a^*}{a_a^*}$$

A restrição de alocação do fator trabalho ($L^* = L_m^* + L_a^*$) determina a fronteira de possibilidades de produção:

$$a_m^* Q_m^* + a_a^* Q_a^* \le L^*$$

Substituindo os valores ($L^*  = 800$, $a_m^*  = 5$ e $a_a^*  = 1$):

$$Q_m^*  = \frac{L_m^*}{5} \quad \text{e} \quad Q_a = \frac{L_a^*}{1}, \quad \text{tal que } L_m^*  + L_a^*  \le 800$$

Produção máxima de manufaturas ($L_m^*  = 800$ e $L_a^*  = 0$):

$$Q_m^*  = \frac{800}{5} \quad \text{e} \quad Q_a = \frac{0}{1}, \quad \text{tal que } L_m^*  + L_a^*  \le 800$$

$$Q_m^*  = 40 \quad \text{e} \quad Q_a = 0$$

Produção máxima de agriculttura ($L_m^*  = 0$ e $L_a^*  = 800$):

$$Q_m^*  = \frac{0}{5} \quad \text{e} \quad Q_a = \frac{800}{1}, \quad \text{tal que } L_m^*  + L_a^*  \le 800$$

$$Q_m^*  = 0 \quad \text{e} \quad Q_a = 800$$

A representação gráfica da fronteira de possibilidades de produção é uma linha reta que une os eixos da manufatura (eixo horizontal) e da agricultura (eixo vertical), cuja inclinação equivale à relação entre as necessidades unitárias de trabalho:

* **Inclinação no país local:** $$\frac{a_m}{a_a} = \frac{3}{2} = 1,5$$
* **Inclinação no país estrangeiro:** $$\frac{a_m^*}{a_a^*} = \frac{5^*}{1^*} = 5^*$$

A inclinação corresponde ao custo de oportunidade oriundo da alocação de trabalho entre os setores. O fato de o trabalho possuir livre mobilidade e constituir o único fator produtivo fundamenta uma fronteira de possibilidades de produção em formato de linha reta negativamente inclinada.

Identificamos que o país local possui vantagem comparativa em relação ao país estrangeiro na produção de manufaturas, enquanto o país estrangeiro possui vantagem comparativa na produção de bens agrícolas. Observe que quando mais unidades de trabalho requedidas, menor a eficiência na produção do bem.



```python
# 1.a

import pandas as pd
import numpy as np
import plotly.graph_objects as go

# Função de produção

def calcular_producao_local(Lm, La, a_m=3, a_a=2):
    """
    Calcula a produção do País Local para alocações de trabalho Lm e La.
    """
    Qm = Lm / a_m
    Qa = La / a_a
    return Qm, Qa

def calcular_producao_estrangeiro(Lm_star, La_star, a_m_star=5, a_a_star=1):
    """
    Calcula a produção do País Estrangeiro para alocações de trabalho Lm* e La*.
    """
    Qm_star = Lm_star / a_m_star
    Qa_star = La_star / a_a_star
    return Qm_star, Qa_star

# Parâmetros
L_local, a_m_local, a_a_local = 1200, 3, 2
L_estrangeiro, a_m_estrangeiro, a_a_estrangeiro = 800, 5, 1

# Pontos de produção máxima no País Local
Qm_max_local, _ = calcular_producao_local(L_local, 0, a_m_local, a_a_local)
_, Qa_max_local = calcular_producao_local(0, L_local, a_m_local, a_a_local)

# Pontos de produção máxima no País Estrangeiro
Qm_max_estrangeiro, _ = calcular_producao_estrangeiro(L_estrangeiro, 0, a_m_estrangeiro, a_a_estrangeiro)
_, Qa_max_estrangeiro = calcular_producao_estrangeiro(0, L_estrangeiro, a_m_estrangeiro, a_a_estrangeiro)

# Tabela com a produção nacional e mundial

dados_producao = {
    "País": ["Local", "Estrangeiro", "Mundial (Máximo)"],
    "Trabalho Total (L)": [L_local, L_estrangeiro, L_local + L_estrangeiro],
    "Prod. Máx. Manufaturas (Qm)": [Qm_max_local, Qm_max_estrangeiro, Qm_max_local + Qm_max_estrangeiro],
    "Prod. Máx. Agricultura (Qa)": [Qa_max_local, Qa_max_estrangeiro, Qa_max_local + Qa_max_estrangeiro],
    "Custo de Oportunidade (a_m / a_a)": [a_m_local / a_a_local, a_m_estrangeiro / a_a_estrangeiro, None]
}

df_producao = pd.DataFrame(dados_producao)

print("Tabela de Capacidade Produtiva:")
print(df_producao.to_string(index=False))

# Gráfico da fronteira de possibilidades de produção

# Vetores de produção para os eixos das FPPs
qm_local_vetor = np.array([0, Qm_max_local])
qa_local_vetor = np.array([Qa_max_local, 0])

qm_estrangeiro_vetor = np.array([0, Qm_max_estrangeiro])
qa_estrangeiro_vetor = np.array([Qa_max_estrangeiro, 0])

# Construção da figura interativa com Plotly
fig = go.Figure()

# Linha da FPP Local
fig.add_trace(go.Scatter(
    x=qm_local_vetor,
    y=qa_local_vetor,
    mode='lines+markers',
    name='FPP País Local',
    line=dict(color='blue', width=3),
    hovertemplate='<b>Local</b><br>Manufaturas: %{x}<br>Agricultura: %{y}<extra></extra>'
))

# Linha da FPP Estrangeiro
fig.add_trace(go.Scatter(
    x=qm_estrangeiro_vetor,
    y=qa_estrangeiro_vetor,
    mode='lines+markers',
    name='FPP País Estrangeiro',
    line=dict(color='red', width=3, dash='dash'),
    hovertemplate='<b>Estrangeiro</b><br>Manufaturas: %{x}<br>Agricultura: %{y}<extra></extra>'
))

# Layout e formatação
fig.update_layout(
    title='Fronteira de Possibilidades de Produção (FPP) : Modelo Ricardiano',
    xaxis_title='Quantidade de Manufaturas (Qm)',
    yaxis_title='Quantidade de Agricultura (Qa)',
    template='plotly_white',
    width=800,
    height=500,
    legend=dict(x=0.7, y=0.95)
)

fig.show()
```

**1.b)** Defina o salário dos setores de manufaturas e de agricultura para cada país (local e estrangeiro) em autarquia. Qual é o papel do salário ($w$)? Apresente a condição de equilíbrio para a produção simultânea de ambos os bens.

Consideramos o mercado competitivo. Portanto, o salário $w_i$ de cada setor depende do produto marginal do trabalho ($\text{PMgL}_i$) e do preço do bem ($P_i$). Assim, temos:

$$w_i = \text{PMgL}_i \cdot P_i$$

Como o produto marginal ($\text{PMgL}_i$) é constante e indicada por:

$$\text{PMgL}_i = \frac{1}{a_i}$$

Rearranjando os temos, o salário pode ser reescrito como:

$$w_i = \frac{P_i}{a_i}$$

Para que a alocação de trabalho permita a produção de ambos os bens, o salário entre os setores deve ser equalizado ($w_i = w_j$). Supomos que deve haver um vetor de preços ($P$) que permita a produção de ambos os bens.

$$w_i = \frac{P_i}{a_i} \quad \text{e} \quad w_j = \frac{P_j}{a_j}$$

---

Isolando a relação entre preços relativos e unidades necessárias de trabalho:

$$\frac{P_i}{P_j} = \frac{a_i}{a_j}$$

A partir dessa relação genérica, podemos responder o item b, estabelecemos o salário local e estrangeiro para manufaturas ($w_m$) e agricultura ($w_a$). Assim, temos:

**Salários no país local:**

* Manufaturas: $$w_m = \text{PMgL}_m \cdot P_m$$
* Agricultura: $$w_a = \text{PMgL}_a \cdot P_a$$

Como podemos substituir pela unidade necessárias de trabalho:

* Manufaturas: $w_m = \frac{P_m}{a_m}$, tal que $\text{PMgL}_m = \frac{1}{a_m}$
* Agricultura: $w_a = \frac{P_a}{a_a}$, tal que $\text{PMgL}_a = \frac{1}{a_a}$

Condição de equilíbrio ($w_m = w_a$). Substituindo:

$$\frac{P_m}{a_m} = \frac{P_a}{a_a}$$

Rearranjando os termos:

$$\frac{P_m}{P_a} = \frac{a_m}{a_a}$$

**Salários no país estrangeiro:**

* Manufaturas: $w_m^* = \text{PMgL}_m^* \cdot P_m^*$
* Agricultura: $w_a^* = \text{PMgL}_a^* \cdot P_a^*$

Como podemos substituir pela necessidade unitária de trabalho:

* Manufaturas: $w_m^* = \frac{P_m^*}{a_m^*}$, tal que $\text{PMgL}_m^* = \frac{1}{a_m^*}$
* Agricultura: $w_a^* = \frac{P_a^*}{a_a^*}$, tal que $\text{PMgL}_a^* = \frac{1}{a_a^*}$

Condição de equilíbrio ($w_m^* = w_a^*$). Substituindo:

$$\frac{P_m^*}{a_m^*} = \frac{P_a^*}{a_a^*}$$

Rearranjando os termos:

$$\frac{P_m^*}{P_a^*} = \frac{a_m^*}{a_a^*}$$

---

Substituindo os valores das necessidades unitárias de trabalho ($a_m = 3$, $a_a = 2$, $a_m^* = 5$ e $a_a^* = 1$), os salários em função dos preços são:

**País Local:**

* Manufaturas: $w_m = \frac{P_m}{3}$
* Agricultura: $w_a = \frac{P_a}{2}$

Condição de equilíbrio ($w_m = w_a$):

$$\frac{P_m}{3} = \frac{P_a}{2} \quad \Rightarrow \quad \frac{P_m}{P_a} = \frac{3}{2}$$

**País Estrangeiro:**

* Manufaturas: $w_m^* = \frac{P_m^*}{5}$
* Agricultura: $w_a^* = \frac{P_a^*}{1} = P_a^*$

Condição de equilíbrio ($w_m^* = w_a^*$):

$$\frac{P_m^*}{5} = \frac{P_a^*}{1} \quad \Rightarrow \quad \frac{P_m^*}{P_a^*} = 5$$

Obs.: Os salários estão em função do preço. Sem o preço absoluto, não é possível identificar o salário nominal absoluto, e a alocação ótima de trabalho depende das preferências da demanda. No entanto, os preços relativos e o custo de oportunidade devem ser iguais para que ambos os bens sejam produzidos simultaneamente.

```python
# 1.b)

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Parâmetros e condições de equilíbrio

# Necessidades unitárias de trabalho (a_i)
a_m_local, a_a_local = 3, 2
a_m_estrangeiro, a_a_estrangeiro = 5, 1

# Para resolver o sistema e encontrar o salário nominal, definimos o bem agrícola 
# como numéraire, fixando seu preço em 1 (Pa = 1).
Pa_local = 1.0
Pa_estrangeiro = 1.0

# Em autarquia, a condição de equilíbrio exige que w_m = w_a.
# Portanto, Pm/Pa = a_m/a_a. Calculamos o preço de autarquia (Pm):
Pm_local = Pa_local * (a_m_local / a_a_local)
Pm_estrangeiro = Pa_estrangeiro * (a_m_estrangeiro / a_a_estrangeiro)

# Cálculo dos salários nominais de equilíbrio (w = P/a)
w_m_local = Pm_local / a_m_local
w_a_local = Pa_local / a_a_local

w_m_estrangeiro = Pm_estrangeiro / a_m_estrangeiro
w_a_estrangeiro = Pa_estrangeiro / a_a_estrangeiro

# 2. Tabela

dados = {
    "País": ["Local", "Local", "Estrangeiro", "Estrangeiro"],
    "Setor": ["Manufaturas", "Agricultura", "Manufaturas", "Agricultura"],
    "Necessidade Unitária (a_i)": [a_m_local, a_a_local, a_m_estrangeiro, a_a_estrangeiro],
    "PMgL (1/a_i)": [1/a_m_local, 1/a_a_local, 1/a_m_estrangeiro, 1/a_a_estrangeiro],
    "Preço (P_i) simulado": [Pm_local, Pa_local, Pm_estrangeiro, Pa_estrangeiro],
    "Salário (w_i)": [w_m_local, w_a_local, w_m_estrangeiro, w_a_estrangeiro],
    "Preço Relativo (Pm/Pa)": [Pm_local/Pa_local, Pm_local/Pa_local, Pm_estrangeiro/Pa_estrangeiro, Pm_estrangeiro/Pa_estrangeiro]
}

df_salarios = pd.DataFrame(dados)

# Arredondando valores para melhor visualização
df_salarios = df_salarios.round(3)

print("Tabela de Equilíbrio em Autarquia (Assumindo Pa = 1 como numéraire):")
print(df_salarios.to_string(index=False))

# 3. Gráfico e condição de equilíbrio

# Vetor de preços relativos (Pm) para simular as funções de salário, dado Pa=1
pm_vetor_local = np.linspace(0, 3, 100)
pm_vetor_estrangeiro = np.linspace(0, 7, 100)

# Funções de salário em relação ao preço da manufatura
w_m_func_local = pm_vetor_local / a_m_local
w_a_func_local = np.full_like(pm_vetor_local, Pa_local / a_a_local) # Linha horizontal (independe de Pm)

w_m_func_est = pm_vetor_estrangeiro / a_m_estrangeiro
w_a_func_est = np.full_like(pm_vetor_estrangeiro, Pa_estrangeiro / a_a_estrangeiro)

# Criando painel com dois gráficos (Subplots)
fig = make_subplots(rows=1, cols=2, subplot_titles=("País Local", "País Estrangeiro"))

# --- Gráfico 1: País Local ---
# Salário Manufatura
fig.add_trace(go.Scatter(x=pm_vetor_local, y=w_m_func_local, mode='lines', name='w_m (Local)', line=dict(color='blue')), row=1, col=1)
# Salário Agricultura
fig.add_trace(go.Scatter(x=pm_vetor_local, y=w_a_func_local, mode='lines', name='w_a (Local)', line=dict(color='green')), row=1, col=1)
# Ponto de Equilíbrio
fig.add_trace(go.Scatter(x=[Pm_local], y=[w_m_local], mode='markers', marker=dict(size=10, color='black'), name='Equilíbrio Local', 
                         hovertemplate='Pm/Pa: %{x}<br>Salário: %{y}<extra></extra>'), row=1, col=1)

# --- Gráfico 2: País Estrangeiro ---
# Salário Manufatura
fig.add_trace(go.Scatter(x=pm_vetor_estrangeiro, y=w_m_func_est, mode='lines', name='w_m* (Estrangeiro)', line=dict(color='red')), row=1, col=2)
# Salário Agricultura
fig.add_trace(go.Scatter(x=pm_vetor_estrangeiro, y=w_a_func_est, mode='lines', name='w_a* (Estrangeiro)', line=dict(color='orange')), row=1, col=2)
# Ponto de Equilíbrio
fig.add_trace(go.Scatter(x=[Pm_estrangeiro], y=[w_m_estrangeiro], mode='markers', marker=dict(size=10, color='black'), name='Equilíbrio Estrangeiro',
                         hovertemplate='Pm*/Pa*: %{x}<br>Salário*: %{y}<extra></extra>'), row=1, col=2)

# Formatação
fig.update_layout(
    title='Determinação do Preço Relativo de Autarquia (Cruzamento dos Salários Setoriais)',
    template='plotly_white',
    width=900,
    height=500
)

# Ajuste dos eixos
fig.update_xaxes(title_text="Preço da Manufatura (Pm)", row=1, col=1)
fig.update_yaxes(title_text="Salário (w)", row=1, col=1)
fig.update_xaxes(title_text="Preço da Manufatura (Pm*)", row=1, col=2)
fig.update_yaxes(title_text="Salário (w*)", row=1, col=2)

fig.show()
```

**d)** Suponha a integração comercial. Formalize o impacto sobre a alocação de trabalho, salário relativo, especialização e padrão de comércio.

**Especialização**

A abertura comercial estabelece um preço relativo mundial $\frac{P_m}{P_a}$ situado entre os custos de oportunidade internos em autarquia:

$$1{,}5 \le \frac{P_m}{P_a} \le 5{,}0$$

Com esse vetor de preços internacionais, a teoria indica a especialização total da economia:

* **País Local:** especializa a totalidade da produção no setor de manufaturas.
* **País Estrangeiro:** especializa a totalidade da produção no setor agrícola.

**Alocação de Trabalho**

A força de trabalho migra totalmente para o setor com vantagem comparativa, garantindo a eficiência produtiva máxima:

* **País Local:** direciona todo o fator trabalho para a manufatura ($L_m = 1200$ e $L_a = 0$). A produção mundial de manufaturas atinge $Q_m = 400$ unidades.
* **País Estrangeiro:** direciona todo o fator trabalho para a agricultura ($L_a^* = 800$ e $L_m^* = 0$). A produção mundial de agricultura atinge $Q_a^* = 800$ unidades.

**Salário Relativo ($\frac{w}{w^*}$)**

Com a integração comercial, cada país determina a remuneração nominal com base no setor em que ocorre a especialização:

* Salário Local (Manufaturas): $w = \frac{P_m}{a_m} = \frac{P_m}{3}$
* Salário Estrangeiro (Agricultura): $w^* = \frac{P_a}{a_a^*} = \frac{P_a}{1}$

A relação entre os salários formaliza a equação do salário relativo:

$$\frac{w}{w^*} = \frac{\frac{P_m}{3}}{\frac{P_a}{1}} = \frac{1}{3} \cdot \left( \frac{P_m}{P_a} \right)$$

Os limites do salário relativo acompanham diretamente as margens dos preços internacionais. Assim, a razão $\frac{w}{w^*}$ varia no intervalo delimitado entre $0{,}5$ (limite inferior, quando $\frac{P_m}{P_a} = 1{,}5$) e aproximadamente $1{,}67$ (limite superior, quando $\frac{P_m}{P_a} = 5{,}0$). A equação confirma que a remuneração reflete a produtividade absoluta do trabalho em conjunto com o preço do bem exportado.

O comércio internacional viabiliza o ganho de bem-estar. As nações superam as restrições impostas pelas antigas fronteiras de possibilidades de produção e ampliam as cestas de consumo disponíveis para a população de modo simultâneo.

**2.a)** Elabore e explique a fronteira de possibilidades de produção para o país Local e Estrangeiro. Considere o cenário de autarquia e a presença de substituição imperfeita de fatores.

A primeira diferença em relação ao modelo ricardiano é que a presença de um fator específico de produção ($K$ ou $T$) acarreta a produtividade marginal decrescente do trabalho. Isto é, as unidades adicionais de trabalho elevam a produção, mas a taxas decrescentes. No limite, o aumento da produção tende a zero. Como consequência, a fronteira de possibilidades de produção é côncava, existindo um ponto ótimo dado o preço relativo e a produtividade marginal relativa do trabalho.

Formalmente:

**Setor $i$:** $\text{PMgL} = \frac{\partial Q}{\partial L} > 0$ e $\frac{\partial \text{PMgL}}{\partial L} = \frac{\partial^2 Q}{\partial L^2} < 0$

Para setor de manufaturas e agricultura, temos:

* **Setor de Manufaturas**: $\text{PMgL}_m = \frac{\partial Q_m}{\partial L_m} > 0$ e $\frac{\partial \text{PMgL}_m}{\partial L_m} = \frac{\partial^2 Q_m}{\partial L_m^2} < 0$
* **Setor de Agricultura**: $\text{PMgL}_a = \frac{\partial Q_a}{\partial L_a} > 0$ e $\frac{\partial \text{PMgL}_a}{\partial L_a} = \frac{\partial^2 Q_a}{\partial L_a^2} < 0$

Obs. Consideramos a funão de produção para cada setor igual, pois a tecnologia é livre. Sendo esperado que cada país adote a tecnologia mais eficiente.

**Alocação Ótima e Maximização do Valor da Produção (Autarquia)**

No modelo de fatores específicos, a "produção máxima" em autarquia refere-se à alocação de trabalho que maximiza o valor total da produção. Isso ocorre no ponto em que o valor da produtividade marginal do trabalho se iguala entre os setores ($P_m \cdot \text{PMgL}_m = P_a \cdot \text{PMgL}_a$).

**País Local:**

* $\text{PMgL}_m = 0,5 \cdot 16^{0,5} \cdot L_m^{-0,5} = 2L_m^{-0,5}$
* $\text{PMgL}_a = 0,5 \cdot 64^{0,5} \cdot L_a^{-0,5} = 4L_a^{-0,5}$

Condição de equilíbrio dado o sistema de preços ($P_m = 1$ e $P_a = 1$):


$$1 \cdot 2L_m^{-0,5} = 1 \cdot 4L_a^{-0,5} \quad \Rightarrow \quad \left(\frac{L_a}{L_m}\right)^{0,5} = 2 \quad \Rightarrow \quad L_a = 4L_m$$

Substituindo na restrição de trabalho ($L_m + L_a = 1000$):


$$L_m + 4L_m = 1000 \quad \Rightarrow \quad L_m = 200 \quad \text{e} \quad L_a = 800$$

A produção ótima que maximiza a alocação em autarquia é:

* **Manufaturas:** $Q_m = 16^{0,5} \cdot 200^{0,5} = 4\sqrt{200} \approx 56,57$
* **Agricultura:** $Q_a = 64^{0,5} \cdot 800^{0,5} = 8\sqrt{800} \approx 226,27$

**País Estrangeiro:**

* $\text{PMgL}_m^* = 0,5 \cdot 64^{0,5} \cdot (L_m^*)^{-0,5} = 4(L_m^*)^{-0,5}$
* $\text{PMgL}_a^* = 0,5 \cdot 16^{0,5} \cdot (L_a^*)^{-0,5} = 2(L_a^*)^{-0,5}$

Condição de equilíbrio dado o sistema de preços ($P_m = 1$ e $P_a = 1$):


$$1 \cdot 4(L_m^*)^{-0,5} = 1 \cdot 2(L_a^*)^{-0,5} \quad \Rightarrow \quad \left(\frac{L_m^*}{L_a^*}\right)^{0,5} = 2 \quad \Rightarrow \quad L_m^* = 4L_a^*$$

Substituindo na restrição de trabalho ($L_m^* + L_a^* = 1000$):


$$4L_a^* + L_a^* = 1000 \quad \Rightarrow \quad L_m^* = 800 \quad \text{e} \quad L_a^* = 200$$

A produção ótima que maximiza a alocação em autarquia é:

* **Manufaturas:** $Q_m^* = 64^{0,5} \cdot 800^{0,5} = 8\sqrt{800} \approx 226,27$
* **Agricultura:** $Q_a^* = 16^{0,5} \cdot 200^{0,5} = 4\sqrt{200} \approx 56,57$

**Produção Máxima Física (Limites da Fronteira de Possibilidades de Produção)**

Caso a análise exija as interceptações da FPP (a capacidade máxima de engenharia de um setor operando de forma extrema, ignorando a demanda e alocando $100\%$ do trabalho disponível), temos:

**País Estrangeiro:**

* Máximo de Agricultura ($L_a^* = 1000$): $Q_a^* = 4\sqrt{1000} \approx 126,49$



```python
# 2.a

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==============================================================================
# 1. PARÂMETROS DO MODELO DE FATORES ESPECÍFICOS (AUTARQUIA)
# ==============================================================================

# Fator móvel (Trabalho)
L_local = 1000
L_estrangeiro = 1000

# Fatores específicos
K_local, T_local = 16, 64
K_estrangeiro, T_estrangeiro = 64, 16

# Sistema de Preços em Autarquia
Pm_aut, Pa_aut = 1, 1

# ==============================================================================
# 2. FUNÇÃO DE RESOLUÇÃO DO EQUILÍBRIO
# ==============================================================================

def calcular_equilibrio(L, K, T, Pm, Pa):
    """
    Calcula a alocação ótima de trabalho e a produção para os setores igualando 
    o valor do produto marginal: Pm * PMgLm = Pa * PMgLa
    """
    # Condição: Pm * 0.5 * K^0.5 * Lm^-0.5 = Pa * 0.5 * T^0.5 * La^-0.5
    # La / Lm = [ (Pa/Pm) * (T^0.5 / K^0.5) ]^2
    razao_La_Lm = ((Pa / Pm) * ((T**0.5) / (K**0.5)))**2
    
    # Substituindo em Lm + La = L
    Lm = L / (1 + razao_La_Lm)
    La = L - Lm
    
    # Produção
    Qm = (K**0.5) * (Lm**0.5)
    Qa = (T**0.5) * (La**0.5)
    
    return Lm, La, Qm, Qa

# Resolução
Lm_loc_aut, La_loc_aut, Qm_loc_aut, Qa_loc_aut = calcular_equilibrio(L_local, K_local, T_local, Pm_aut, Pa_aut)
Lm_est_aut, La_est_aut, Qm_est_aut, Qa_est_aut = calcular_equilibrio(L_estrangeiro, K_estrangeiro, T_estrangeiro, Pm_aut, Pa_aut)

# ==============================================================================
# 3. TABELA DE RESULTADOS
# ==============================================================================

dados = {
    "País": ["Local", "Estrangeiro"],
    "L_m (Trabalho Manuf.)": [Lm_loc_aut, Lm_est_aut],
    "L_a (Trabalho Agric.)": [La_loc_aut, La_est_aut],
    "Q_m (Prod. Manuf.)": [Qm_loc_aut, Qm_est_aut],
    "Q_a (Prod. Agric.)": [Qa_loc_aut, Qa_est_aut]
}

df_resultados = pd.DataFrame(dados)

print("TABELA DE ALOCAÇÃO E PRODUÇÃO EM AUTARQUIA")
print("-" * 70)
print(df_resultados.to_string(index=False, float_format="%.2f"))
print("-" * 70)

# ==============================================================================
# 4. GRÁFICO: FRONTEIRAS DE POSSIBILIDADES DE PRODUÇÃO
# ==============================================================================

# Vetor de alocação de trabalho para desenhar a curva da FPP
lm_vetor = np.linspace(0, L_local, 500)
la_vetor = L_local - lm_vetor

# FPP País Local
qm_fpp_loc = (K_local**0.5) * (lm_vetor**0.5)
qa_fpp_loc = (T_local**0.5) * (la_vetor**0.5)

# FPP País Estrangeiro
qm_fpp_est = (K_estrangeiro**0.5) * (lm_vetor**0.5)
qa_fpp_est = (T_estrangeiro**0.5) * (la_vetor**0.5)

fig = make_subplots(rows=1, cols=2, subplot_titles=("País Local", "País Estrangeiro"))

# --- FPP Local ---
fig.add_trace(go.Scatter(
    x=qm_fpp_loc, y=qa_fpp_loc, mode='lines', name='FPP Local', line=dict(color='blue', width=3)
), row=1, col=1)

# Ponto Autarquia Local
fig.add_trace(go.Scatter(
    x=[Qm_loc_aut], y=[Qa_loc_aut], mode='markers', name='Ponto Ótimo', 
    marker=dict(size=12, color='black', symbol='circle'),
    hovertemplate='Qm: %{x:.1f}<br>Qa: %{y:.1f}<extra></extra>'
), row=1, col=1)


# --- FPP Estrangeiro ---
fig.add_trace(go.Scatter(
    x=qm_fpp_est, y=qa_fpp_est, mode='lines', name='FPP Estrangeiro', line=dict(color='green', width=3)
), row=1, col=2)

# Ponto Autarquia Estrangeiro
fig.add_trace(go.Scatter(
    x=[Qm_est_aut], y=[Qa_est_aut], mode='markers', name='Ponto Ótimo', 
    marker=dict(size=12, color='black', symbol='circle'),
    hovertemplate='Qm*: %{x:.1f}<br>Qa*: %{y:.1f}<extra></extra>'
), row=1, col=2)

# Ajustes de Layout
fig.update_layout(
    title='Fronteira de Possibilidades de Produção e Alocação em Autarquia',
    template='plotly_white',
    width=900, height=500,
    showlegend=False
)

fig.update_xaxes(title_text="Manufaturas (Q_m)", row=1, col=1)
fig.update_yaxes(title_text="Agricultura (Q_a)", range=[0, 300], row=1, col=1)

fig.update_xaxes(title_text="Manufaturas (Q_m*)", row=1, col=2)
fig.update_yaxes(title_text="Agricultura (Q_a*)", range=[0, 300], row=1, col=2)

fig.show()
```

**2.b)** Defina o salário de cada setor em autarquia. Qual é a condição de equilíbrio e o papel da produtividade marginal do trabalho ($PMgL$)?

**Definição do salário em cada setor (Autarquia)**

Em um mercado competitivo, o salário nominal em cada setor ($w_i$) é determinado pelo valor do produto marginal do trabalho, formalizado pelo produto entre o preço do bem e a produtividade marginal do trabalho ($\text{PMgL}_i$):

* **Setor de Manufaturas:** $w_m = P_m \cdot \text{PMgL}_m$
* **Setor de Agricultura:** $w_a = P_a \cdot \text{PMgL}_a$

Como os preços em autarquia são normalizados ($P_m = 1$ e $P_a = 1$), os salários nominais equivalem diretamente às respectivas produtividades marginais.

---

**Condição de equilíbrio**

Como o fator trabalho é móvel entre os setores no modelo de fatores específicos, a concorrência garante a livre mobilidade até que os retornos salariais se igualem. Portanto, a condição de equilíbrio exige a equalização dos salários nominais entre as indústrias:

$$w_m = w_a \quad \Rightarrow \quad P_m \cdot \text{PMgL}_m = P_a \cdot \text{PMgL}_a$$

Essa igualdade orienta a alocação ótima da força de trabalho total entre os setores de manufaturas e de agricultura ($L_m$ e $L_a$).

---

**O papel da produtividade marginal do trabalho ($\text{PMgL}$)**

A $\text{PMgL}$ representa o acréscimo na produção gerado pela contratação de uma unidade adicional de trabalho. No modelo de fatores específicos, a presença de fatores fixos (capital $K$ ou terra $T$) faz com que a $\text{PMgL}$ exiba **rendimentos decrescentes** ($\frac{\partial \text{PMgL}}{\partial L} < 0$).

O papel central da $\text{PMgL}$ consiste em ditar o comportamento decrescente da remuneração conforme o trabalho se expande em um setor, estabelecendo o limite para a alocação de recursos e determinando diretamente a curvatura (concavidade) da Fronteira de Possibilidades de Produção (FPP).

```python
# 2.b

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==============================================================================
# 1. PARÂMETROS E RESOLUÇÃO ANALÍTICA DO EQUILÍBRIO EM AUTARQUIA
# ==============================================================================

# Dotações
L_local, K_local, T_local = 1000, 16, 64
L_estrangeiro, K_estrangeiro, T_estrangeiro = 1000, 64, 16

# Preços em Autarquia (Normalizados)
Pm, Pa = 1.0, 1.0
preco_relativo = Pm / Pa  # 1.0

def resolver_autarquia(L, K, T, Pm, Pa):
    """
    Calcula analiticamente a alocação de trabalho, produção, produtividade marginal
    e salários em autarquia para um país.
    """
    razao_La_Lm = ((Pa / Pm) * ((T**0.5) / (K**0.5)))**2
    
    Lm = L / (1 + razao_La_Lm)
    La = L - Lm
    
    Qm = (K**0.5) * (Lm**0.5)
    Qa = (T**0.5) * (La**0.5)
    
    pmgl_m = 0.5 * (K**0.5) * (Lm**-0.5)
    pmgl_a = 0.5 * (T**0.5) * (La**-0.5)
    
    w_m = Pm * pmgl_m
    w_a = Pa * pmgl_a
    
    return {
        "Lm": Lm, "La": La,
        "Qm": Qm, "Qa": Qa,
        "PMgLm": pmgl_m, "PMgLa": pmgl_a,
        "wm": w_m, "wa": w_a
    }

# Executando para ambos os países
eq_local = resolver_autarquia(L_local, K_local, T_local, Pm, Pa)
eq_estrangeiro = resolver_autarquia(L_estrangeiro, K_estrangeiro, T_estrangeiro, Pm, Pa)

# ==============================================================================
# 2. TABELA DE RESULTADOS ANALÍTICOS
# ==============================================================================

dados_tabela = {
    "Variável / Setor": [
        "Trabalho no Setor (Lm, La)", 
        "Produção Total (Qm, Qa)", 
        "Produtividade Marginal (PMgL)", 
        "Salário Nominal de Equilíbrio (wm = wa)"
    ],
    "País Local - Manufaturas": [
        f"{eq_local['Lm']:.2f}", f"{eq_local['Qm']:.2f}", 
        f"{eq_local['PMgLm']:.4f}", f"{eq_local['wm']:.4f}"
    ],
    "País Local - Agricultura": [
        f"{eq_local['La']:.2f}", f"{eq_local['Qa']:.2f}", 
        f"{eq_local['PMgLa']:.4f}", f"{eq_local['wa']:.4f}"
    ],
    "País Estrangeiro - Manufaturas": [
        f"{eq_estrangeiro['Lm']:.2f}", f"{eq_estrangeiro['Qm']:.2f}", 
        f"{eq_estrangeiro['PMgLm']:.4f}", f"{eq_estrangeiro['wm']:.4f}"
    ],
    "País Estrangeiro - Agricultura": [
        f"{eq_estrangeiro['La']:.2f}", f"{eq_estrangeiro['Qa']:.2f}", 
        f"{eq_estrangeiro['PMgLa']:.4f}", f"{eq_estrangeiro['wa']:.4f}"
    ]
}

df_resultados = pd.DataFrame(dados_tabela)

print("TABELA DE RESULTADOS DO EQUILÍBRIO EM AUTARQUIA")
print("=" * 95)
print(df_resultados.to_string(index=False))
print("=" * 95 + "\n")

# ==============================================================================
# 3. GRÁFICOS: FPP, PONTO DE EQUILÍBRIO E RETA DE PREÇO RELATIVO TANGENTE
# ==============================================================================

fig = make_subplots(
    rows=1, cols=2, 
    subplot_titles=("País Local: FPP e Preço Relativo", "País Estrangeiro: FPP e Preço Relativo")
)

# Vetores para desenhar a FPP
lm_vetor = np.linspace(1, L_local - 1, 500)
la_vetor_loc = L_local - lm_vetor
la_vetor_est = L_estrangeiro - lm_vetor

qm_loc = (K_local**0.5) * (lm_vetor**0.5)
qa_loc = (T_local**0.5) * (la_vetor_loc**0.5)

qm_est = (K_estrangeiro**0.5) * (lm_vetor**0.5)
qa_est = (T_estrangeiro**0.5) * (la_vetor_est**0.5)

# Vetores para desenhar a reta tangente (Preço Relativo Pm/Pa) passando pelo ponto ótimo
# Equação da reta de valor/preço: Qa = Qa_otimo + (Pm/Pa)*(Qm_otimo - Qm)
qm_tangente = np.linspace(0, 250, 100)
qa_tangente_local = eq_local['Qa'] - preco_relativo * (qm_tangente - eq_local['Qm'])
qa_tangente_est = eq_estrangeiro['Qa'] - preco_relativo * (qm_tangente - eq_estrangeiro['Qm'])

# --- Subplot 1: País Local ---
fig.add_trace(go.Scatter(
    x=qm_loc, y=qa_loc, mode='lines', name='FPP Local', 
    line=dict(color='royalblue', width=3)
), row=1, col=1)

# Reta Tangente (Preço Relativo)
fig.add_trace(go.Scatter(
    x=qm_tangente, y=qa_tangente_local, mode='lines', name='Preço Relativo (Pm/Pa)', 
    line=dict(color='crimson', width=2, dash='dash')
), row=1, col=1)

fig.add_trace(go.Scatter(
    x=[eq_local['Qm']], y=[eq_local['Qa']], mode='markers+text', 
    name='Equilíbrio Autarquia (Local)',
    text=[f" Autarquia<br> (Qm={eq_local['Qm']:.1f}, Qa={eq_local['Qa']:.1f})"],
    textposition="top right",
    marker=dict(size=12, color='darkblue', symbol='circle')
), row=1, col=1)

# --- Subplot 2: País Estrangeiro ---
fig.add_trace(go.Scatter(
    x=qm_est, y=qa_est, mode='lines', name='FPP Estrangeiro', 
    line=dict(color='seagreen', width=3)
), row=1, col=2)

# Reta Tangente (Preço Relativo)
fig.add_trace(go.Scatter(
    x=qm_tangente, y=qa_tangente_est, mode='lines', name='Preço Relativo (Pm*/Pa*)', 
    line=dict(color='crimson', width=2, dash='dash')
), row=1, col=2)

fig.add_trace(go.Scatter(
    x=[eq_estrangeiro['Qm']], y=[eq_estrangeiro['Qa']], mode='markers+text', 
    name='Equilíbrio Autarquia (Estrangeiro)',
    text=[f" Autarquia<br> (Qm*={eq_estrangeiro['Qm']:.1f}, Qa*={eq_estrangeiro['Qa']:.1f})"],
    textposition="top right",
    marker=dict(size=12, color='darkgreen', symbol='circle')
), row=1, col=2)

# Configurações de layout
fig.update_layout(
    title='Equilíbrio em Autarquia: FPP e Reta Tangente de Preço Relativo (Pm/Pa)',
    template='plotly_white',
    width=1000, height=550,
    showlegend=True,
    legend=dict(y=-0.2, orientation="h")
)

fig.update_xaxes(title_text="Manufaturas (Q_m)", row=1, col=1)
fig.update_yaxes(title_text="Agricultura (Q_a)", range=[0, 300], row=1, col=1)

fig.update_xaxes(title_text="Manufaturas (Q_m*)", row=1, col=2)
fig.update_yaxes(title_text="Agricultura (Q_a*)", range=[0, 300], row=1, col=2)

fig.show()
```

**3.a)** Elabore e explique a fronteira de possibilidades de produção para o país Local e Estrangeiro. Considere o cenário de autarquia e a presença de substituição imperfeita de fatores.

A Fronteira de Possibilidades de Produção (FPP) possui o formato côncavo, oriundo da substituição imperfeita de fatores (capital e trabalho), dada uma certa tecnologia representada pela função de produção do tipo Cobb-Douglas.

Os parâmetros formados pelos dados sobre dotação de fatores e função de produção permitem identificar dois mecanismos do modelo H-O. Primeiro, o país tende a ser mais eficiente na produção do bem que utiliza intensivamente o seu fator relativamente abundante. Dessa forma, o país local é mais eficiente na produção de bens agrícolas, enquanto o país estrangeiro é na produção de manufaturas. Segundo, essa eficiência deriva do fato de a abundância do fator criar incentivos para maior produção do bem intensivo nesse insumo, dada a função de produção e a tecnologia comum entre os dois países.

Vale lembrar que os mercados são perfeitamente concorrenciais, a tecnologia é idêntica e de livre acesso entre os países, e há perfeita mobilidade interna dos fatores de produção.

```python
# 3.a

import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

L_loc, K_loc = 1200, 400
L_for, K_for = 400, 1200

def gerar_fpp(L_total, K_total, n_pontos=1000):
    L_m = np.linspace(1e-4 * L_total, (1 - 1e-4) * L_total, n_pontos)
    K_m = (9 * L_m * K_total) / (L_total + 8 * L_m)
    
    L_a = L_total - L_m
    K_a = K_total - K_m
    
    Q_m = (K_m ** 0.75) * (L_m ** 0.25)
    Q_a = (K_a ** 0.25) * (L_a ** 0.75)
    return Q_m, Q_a

Qm_loc, Qa_loc = gerar_fpp(L_loc, K_loc)
Qm_for, Qa_for = gerar_fpp(L_for, K_for)

fig, ax = plt.subplots(1, 2, figsize=(12, 6), dpi=300)

# --- PAÍS LOCAL ---
ax[0].plot(Qm_loc, Qa_loc, color='#0d5c91', linewidth=2.8)
ax[0].fill_between(Qm_loc, Qa_loc, color='#0d5c91', alpha=0.12)
ax[0].scatter([0, Qm_loc[-1]], [Qa_loc[0], 0], color='#0d5c91', s=50, zorder=5)

qa_max_loc = Qa_loc[0]
qm_max_loc = Qm_loc[-1]

ax[0].text(30, qa_max_loc - 40, f'$Q_a = {qa_max_loc:.1f}$', fontsize=10, fontweight='bold', color='#0d5c91')
ax[0].text(qm_max_loc - 220, 40, f'$Q_m = {qm_max_loc:.1f}$', fontsize=10, fontweight='bold', color='#0d5c91')

ax[0].set_title('País Local (Abundante em Trabalho)\n$L = 1200,\\, K = 400$', fontsize=11, pad=10, fontweight='bold')
ax[0].set_xlabel('Manufaturas ($Q_m$)', fontsize=10, labelpad=6)
ax[0].set_ylabel('Agricultura ($Q_a$)', fontsize=10, labelpad=6)
ax[0].set_xlim(-20, 1000)
ax[0].set_ylim(-20, 1000)
ax[0].set_aspect('equal', adjustable='box')
ax[0].grid(True, linestyle=':', alpha=0.6)

# --- PAÍS ESTRANGEIRO ---
ax[1].plot(Qm_for, Qa_for, color='#b30000', linewidth=2.8)
ax[1].fill_between(Qm_for, Qa_for, color='#b30000', alpha=0.12)
ax[1].scatter([0, Qm_for[-1]], [Qa_for[0], 0], color='#b30000', s=50, zorder=5)

qa_max_for = Qa_for[0]
qm_max_for = Qm_for[-1]

ax[1].text(30, qa_max_for + 30, f'$Q_a^* = {qa_max_for:.1f}$', fontsize=10, fontweight='bold', color='#b30000')
ax[1].text(qm_max_for - 220, 40, f'$Q_m^* = {qm_max_for:.1f}$', fontsize=10, fontweight='bold', color='#b30000')

ax[1].set_title('País Estrangeiro (Abundante em Capital)\n$L^* = 400,\\, K^* = 1200$', fontsize=11, pad=10, fontweight='bold')
ax[1].set_xlabel('Manufaturas ($Q_m$)', fontsize=10, labelpad=6)
ax[1].set_ylabel('Agricultura ($Q_a$)', fontsize=10, labelpad=6)
ax[1].set_xlim(-20, 1000)
ax[1].set_ylim(-20, 1000)
ax[1].set_aspect('equal', adjustable='box')
ax[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.savefig('fpp_heckscher_ohlin_calibrado.png', dpi=300)
```

**3.b)** Identifique o ponto ótimo de produção dos dois bens para cada país em autarquia. Represente o equilíbrio graficamente.

O equilíbrio da alocação de fatores e nível de produção consiste na taxa marginal de substituição técnica ($TMST_m = TMST_a$) e a remunerçaração relativa dos fatores ($\frac{p_m}{p_a}$). Para isso, seguimos o roteiro:

1. Produtividades Marginais e Curva de Contrato

A eficiência na alocação interna exige a equalização das Taxas Marginais de Substituição Técnica ($TMST_m = TMST_a$):

* Manufaturas ($Q_m = K_m^{0,75} L_m^{0,25}$):$$PMgL_m = \frac{\partial Q_m}{\partial L_m} = 0,25 \cdot K_m^{0,75} L_m^{-0,75}$$$$PMgK_m = \frac{\partial Q_m}{\partial K_m} = 0,75 \cdot K_m^{-0,25} L_m^{0,25}$$

$$TMST_m = \frac{PMgL_m}{PMgK_m} = \frac{0,25}{0,75} \frac{K_m}{L_m} = \frac{1}{3} \frac{K_m}{L_m}$$Agricultura ($Q_a = K_a^{0,25} L_a^{0,75}$):$$PMgL_a = \frac{\partial Q_a}{\partial L_a} = 0,75 \cdot K_a^{0,25} L_a^{-0,25}$$

$$PMgK_a = \frac{\partial Q_a}{\partial K_a} = 0,25 \cdot K_a^{-0,75} L_a^{0,75}$$
$$TMST_a = \frac{PMgL_a}{PMgK_a} = \frac{0,75}{0,25} \frac{K_a}{L_a} = 3 \frac{K_a}{L_a}$$

Igualando $TMST_m = TMST_a$ (Curva de Contrato):$$\frac{1}{3} \frac{K_m}{L_m} = 3 \frac{K_a}{L_a} \implies \frac{K_a}{L_a} = \frac{1}{9} \frac{K_m}{L_m}$$

2. Relação entre Preços Relativos e Intensidade de Capital ($K_m/L_m$)

Em mercados competitivos, a Taxa Marginal de Transformação ($RMT$) iguala o preço relativo dos bens ($P_m/P_a$):

$$RMT = \frac{PMgL_a}{PMgL_m} = \frac{0,75 \left(\frac{K_a}{L_a}\right)^{0,25}}{0,25 \left(\frac{K_m}{L_m}\right)^{0,75}} = \frac{P_m}{P_a}$$

Substituindo a relação da curva de contrato ($\frac{K_a}{L_a} = \frac{1}{9} \frac{K_m}{L_m}$):

$$\frac{0,75 \left(\frac{1}{9} \frac{K_m}{L_m}\right)^{0,25}}{0,25 \left(\frac{K_m}{L_m}\right)^{0,75}} = \frac{P_m}{P_a}$$$$3 \cdot \left(\frac{1}{9}\right)^{0,25} \cdot \frac{\left(\frac{K_m}{L_m}\right)^{0,25}}{\left(\frac{K_m}{L_m}\right)^{0,75}} = \frac{P_m}{P_a}$$

Como $\left(\frac{1}{9}\right)^{0,25} = \frac{1}{\sqrt{3}}$:$$3 \cdot \frac{1}{\sqrt{3}} \cdot \frac{1}{\sqrt{\frac{K_m}{L_m}}} = \frac{P_m}{P_a} \implies \frac{\sqrt{3}}{\sqrt{\frac{K_m}{L_m}}} = \frac{P_m}{P_a}$$

Elevando ambos os lados ao quadrado:

$$\frac{3}{\frac{K_m}{L_m}} = \left(\frac{P_m}{P_a}\right)^2 \implies \frac{K_m}{L_m} = 3 \left(\frac{P_a}{P_m}\right)^2$$

Essa é a Equação 1 do sistema.

3. Alocação

$$L_a = L - L_m$$$$K_a = K - K_m$$

Da Equação 1, sabe-se que $K_m = 3 \left(\frac{P_a}{P_m}\right)^2 L_m$. 

Substituindo na razão de fatores do setor agrícola ($\frac{K_a}{L_a} = \frac{1}{9} \frac{K_m}{L_m}$):

$$\frac{K - K_m}{L - L_m} = \frac{1}{3} \left(\frac{P_a}{P_m}\right)^2$$

Substituindo $K_m = 3 \left(\frac{P_a}{P_m}\right)^2 L_m$:

$$\frac{K - 3 \left(\frac{P_a}{P_m}\right)^2 L_m}{L - L_m} = \frac{1}{3} \left(\frac{P_a}{P_m}\right)^2$$

Multiplicando cruzado:$$3 K - 9 \left(\frac{P_a}{P_m}\right)^2 L_m = \left(\frac{P_a}{P_m}\right)^2 L - \left(\frac{P_a}{P_m}\right)^2 L_m$$

Isolando os termos com $L_m$ no lado direito:

$$3 K - \left(\frac{P_a}{P_m}\right)^2 L = 8 \left(\frac{P_a}{P_m}\right)^2 L_m$$

Isolando $L_m$:$$L_m = \frac{3 K - \left(\frac{P_a}{P_m}\right)^2 L}{8 \left(\frac{P_a}{P_m}\right)^2}$$

Dividindo o numerador e o denominador por $\left(\frac{P_a}{P_m}\right)^2$, obtém-se a forma reduzida direta:

$$L_m = \frac{3}{8} K \left(\frac{P_m}{P_a}\right)^2 - \frac{L}{8}$$

-------

**Aplicação: país local**

Caso 1: Preço Relativo $P_m/P_a = 1,0$

Alocação de Trabalho na Indústria ($L_m$):

$$L_m = \frac{3}{8}(400)(1,0)^2 - \frac{1200}{8} = 150(1,0) - 150 = 150 - 150 = 0$$

Alocação de Capital na Indústria ($K_m$):$$K_m = 0 \cdot 3 \cdot \left(\frac{1}{1,0}\right)^2 = 0$$

Fatores Remanescentes na Agricultura ($L_a$ e $K_a$):

$$L_a = L - L_m = 1200 - 0 = 1200$$

$$K_a = K - K_m = 400 - 0 = 400$$

Produção Resultante:$$Q_m = (0)^{0,75} (0)^{0,25} = 0$$

$$Q_a = (400)^{0,25} (1200)^{0,75} = 400 \cdot 3^{0,75} \approx 911,8$$

Caso 2: Preço Relativo $P_m/P_a = 1,5$

Alocação de Trabalho na Indústria ($L_m$):$$L_m = \frac{3}{8}(400)(1,5)^2 - \frac{1200}{8} = 150(2,25) - 150 = 337,5 - 150 = 187,5$$

Alocação de Capital na Indústria ($K_m$):

$$K_m = 187,5 \cdot 3 \cdot \left(\frac{2}{3}\right)^2 = 187,5 \cdot 3 \cdot \frac{4}{9} = 187,5 \cdot \frac{4}{3} = 250,0$$

Fatores Remanescentes na Agricultura ($L_a$ e $K_a$):$$L_a = 1200 - 187,5 = 1012,5$$

$$K_a = 400 - 250,0 = 150,0$$

Produção Resultante:

$$Q_m = (250,0)^{0,75} (187,5)^{0,25} \approx 232,6$$
$$Q_a = (150,0)^{0,25} (1012,5)^{0,75} \approx 629,0$$

---
**Aplicação: país estrangeiro**

Caso 1: Preço Relativo $P_m^*/P_a^* = 1,0$

Alocação de Trabalho na Indústria ($L_m^*$):

$$L_m^* = \frac{3}{8}(1200)(1,0)^2 - \frac{400}{8} = 450(1,0) - 50 = 450 - 50 = 400$$

Como $L_m^* = 400 = L^*$, a totalidade da força de trabalho do País Estrangeiro aloca na indústria.Alocação de Capital na Indústria ($K_m^*$):

$$K_m^* = 400 \cdot 3 \cdot \left(\frac{1}{1,0}\right)^2 = 1200$$

Como $K_m^* = 1200 = K^*$, a totalidade do capital do País Estrangeiro aloca na indústria.Fatores Remanescentes na Agricultura ($L_a^*$ e $K_a^*$):$$L_a^* = L^* - L_m^* = 400 - 400 = 0$$$$K_a^* = K^* - K_m^* = 1200 - 1200 = 0$$

Produção Resultante:

$$Q_m^* = (1200)^{0,75} (400)^{0,25} = 400 \cdot 3^{0,75} \approx 911,8$$$$Q_a^* = (0)^{0,25} (0)^{0,75} = 0$$

O País Estrangeiro opera na borda oposta ao País Local, especializando totalmente na produção do bem intensivo em capital.

Caso 2: Preço Relativo $P_m^*/P_a^* = 1,5$

Alocação de Trabalho na Indústria ($L_m^*$):$$L_m^* = \frac{3}{8}(1200)(1,5)^2 - \frac{400}{8} = 450(2,25) - 50 = 1012,5 - 50 = 962,5$$

Como $L_m^* = 962,5$ excede a dotação total de trabalho do país ($L^* = 400$), a economia atinge a restrição física de pleno emprego e permanece no ponto de canto de máxima especialização industrial:Alocação efetiva: $L_m^* = 400$, $K_m^* = 1200$, $L_a^* = 0$, $K_a^* = 0$Produção efetiva: $Q_m^* \approx 911,8$ e $Q_a^* = 0$

Caso Espelho Simétrico: Preço Relativo $P_m^*/P_a^* = 2/3 \approx 0,67$Como o País Estrangeiro é abundante em capital, o preço relativo interno que produz o ponto interior simétrico ao País Local (onde $P_m/P_a = 1,5$) é o preço inverso $P_m^*/P_a^* = \frac{1}{1,5} = \frac{2}{3} \approx 0,67$:

Alocação de Trabalho na Indústria ($L_m^*$):

$$L_m^* = \frac{3}{8}(1200)\left(\frac{2}{3}\right)^2 - \frac{400}{8} = 450 \left(\frac{4}{9}\right) - 50 = 200 - 50 = 150,0$$

Alocação de Capital na Indústria ($K_m^*$):

$$K_m^* = 150,0 \cdot 3 \cdot \left(\frac{3}{2}\right)^2 = 150,0 \cdot 3 \cdot \frac{9}{4} = 150,0 \cdot \frac{27}{4} = 1012,5$$

Fatores Remanescentes na Agricultura ($L_a^*$ e $K_a^*$):

$$L_a^* = 400 - 150,0 = 250,0$$
$$K_a^* = 1200 - 1012,5 = 187,5$$

Produção Resultante:

$$Q_m^* = (1012,5)^{0,75} (150,0)^{0,25} \approx 628,2$$
$$Q_a^* = (187,5)^{0,25} (250,0)^{0,75} \approx 232,7$$

Obs.: o resultado é simétrico devido ao país local ser o espelho inversos do estrangeiro. Foi uma simplificação. Não precisam ser iguais.
