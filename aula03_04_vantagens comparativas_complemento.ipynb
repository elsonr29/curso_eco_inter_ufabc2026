#### **2. Economia de fator único: autarquia**

Para desenvolver o modelo ricardiano e a importância das vantagens comparativas, construímos uma economia hipotética em autarquia:

* Mercado competitivo;
* A economia produz e consome dois bens ($q_1$ e $q_2$);
* O único fator de produção é o trabalho ($L$), com cada setor apresentando uma necessidade unitária de trabalho que indica a eficiência ($a_1$ e $a_2$);
* A alocação do trabalho depende dos salários ($w_1$ e $w_2$), que são influenciados pelos preços relativos dos bens ($p_1$ e $p_2$).

**Importante:** A economia em autarquia deve apresentar um salário de equilíbrio ($w^*$), no qual os salários dos dois setores são iguais ($w_1=w_2$). Assim, ocorre a produção dos dois bens.



**2.1. Alocação de trabalho**

A alocação de trabalho ($L$) entre os setores determina o nível de produção de cada bem ($q_1$ e $q_2$), dada a necessidade unitária de trabalho correspondente ($a_1$ e $a_2$).

- Alocação de trabalho:

$$ L_1 + L_2 \leq L \tag{1} $$

- Necessidade unitária de trabalho:

$$ a_{1} q_{1} = L_{1} \quad (\text{Bem 1}) \tag{2} $$

$$ a_{2} q_{2} = L_{2} \quad (\text{Bem 2}) \tag{3} $$

- Podemos reescrever (1) utilizando (2) e (3):

$$ a_{1} q_{1} + a_{2} q_{2} \leq L \tag{4} $$


```python
# 2.1. Alocação de trabalho

import numpy as np
import pandas as pd

class EconomiaRicardiana:
    def __init__(self, trabalho_total, a1, a2):
        self.L = trabalho_total
        self.a1 = a1
        self.a2 = a2

    def gerar_alocacoes(self, num_pontos=100):
        # Produção máxima do bem 1
        q1_max = self.L / self.a1
        
        # Vetor com diferentes níveis de produção do bem 1
        q1_vetor = np.linspace(0, q1_max, num_pontos)
        
        # Alocação de trabalho para o setor 1 (L1 = a1 * q1)
        L1_vetor = self.a1 * q1_vetor
        
        # Trabalho restante destinado ao setor 2 (L2 = L - L1)
        L2_vetor = self.L - L1_vetor
        
        # Produção do bem 2 (q2 = L2 / a2)
        q2_vetor = L2_vetor / self.a2
        
        # Organização dos dados
        dados = pd.DataFrame({
            'L1': L1_vetor,
            'L2': L2_vetor,
            'q1': q1_vetor,
            'q2': q2_vetor
        })
        
        return dados
```

**2.2. Fronteira de Possibilidades de Produção**

A partir da alocação de trabalho e da necessidade unitária de trabalho, podemos desenhar a Fronteira de Possibilidades de Produção (FPP). A FPP relaciona a mudança na alocação do fator trabalho com a produção dos dois bens ($q_1$ e $q_2$).

Além disso, identificamos o custo de oportunidade, dado pela relação entre a necessidade unitária de trabalho do bem 1 ($a_1$) e do bem 2 ($a_2$). Essa relação indica o custo para produzir uma unidade adicional do bem 1 em termos do bem 2.

$$ \text{Custo de Oportunidade} = \frac{a_1}{a_2} \tag{5} $$

O custo de oportunidade é igual à inclinação da Fronteira de Possibilidades de Produção.

```python
# 2.2. Fronteira de Possibilidades de Produção (FPP)

import matplotlib.pyplot as plt

fpp_dados = {
    'q1': [0, 10],
    'q2': [10, 0]
}

# Configuração e geração do gráfico da FPP
plt.figure(figsize=(8, 6))

# Traço da fronteira em cor neutra para visual acadêmico
plt.plot(fpp_dados['q1'], fpp_dados['q2'], color='black', linewidth=2)
plt.fill_between(fpp_dados['q1'], fpp_dados['q2'], color='lightgray', alpha=0.4)

# Configuração dos eixos para exibir apenas letras (variáveis algébricas)
# O Matplotlib vai pegar o número 10 e esconder, exibindo apenas o LaTeX
plt.xticks([max(fpp_dados['q1'])], [r'$L/a_1$'], fontsize=14)
plt.yticks([max(fpp_dados['q2'])], [r'$L/a_2$'], fontsize=14)

# Rótulos
plt.xlabel('Produção do Bem 1 ($q_1$)', fontsize=12)
plt.ylabel('Produção do Bem 2 ($q_2$)', fontsize=12)

# Ajuste dos limites dos eixos
plt.xlim(0, max(fpp_dados['q1']) * 1.1)
plt.ylim(0, max(fpp_dados['q2']) * 1.1)

# Remoção das bordas superior e direita para estética teórica
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

**2.4. Preços relativos e salários**

No modelo ricardiano, o salário em cada setor iguala o valor do produto marginal do trabalho. Como a produtividade marginal do trabalho é constante e igual a $1/a_i$, o salário do setor $i$ é dado por:

$$ w_i = \frac{p_i}{a_i} \tag{6} $$

Para os setores dos bens 1 e 2, temos:

$$ w_1 = \frac{p_1}{a_1} \tag{7} $$

$$ w_2 = \frac{p_2}{a_2} \tag{8} $$

Na condição de equilíbrio, os salários dos dois setores devem ser iguais ($w_1 = w_2$). Portanto, podemos igualar (7) e (8) para identificar o equilíbrio:

$$ \frac{p_1}{a_1} = \frac{p_2}{a_2} \tag{9} $$

A reorganização da equação (8) permite separar o preço relativo $\left(\frac{p_1}{p_2}\right)$ e o custo de oportunidade $\left(\frac{a_1}{a_2}\right)$:

$$\frac{p_1}{p_2} = \frac{a_1}{a_2} \tag{9} $$


**Importante:** O mercado de trabalho é competitivo, sem imperfeições ou flutuações de curto prazo. Portanto, o salário reflete o produto marginal gerado pelo trabalhador.

```python
# 2.4 preços relativos e salários: setar

import numpy as np

class MercadoTrabalhoRicardiano:
    def __init__(self, a1, a2, p1, p2):
        self.a1 = a1
        self.a2 = a2
        self.p1 = p1
        self.p2 = p2

    def calcular_salarios(self):
        # Aplicação das equações (7) e (8)
        w1 = self.p1 / self.a1
        w2 = self.p2 / self.a2
        return w1, w2

    def analisar_equilibrio(self):
        # Definição das variáveis da equação (9)
        preco_relativo = self.p1 / self.p2
        custo_oportunidade = self.a1 / self.a2
        
        # Comparação para identificar a alocação do trabalho
        if preco_relativo > custo_oportunidade:
            return "Especialização total no Bem 1 (w1 > w2)"
        elif preco_relativo < custo_oportunidade:
            return "Especialização total no Bem 2 (w2 > w1)"
        else:
            return "Equilíbrio: Produção simultânea de ambos os bens (w1 = w2)"


```

```python
# 2.4 preços relativos e salários: gráficos

import matplotlib.pyplot as plt
import numpy as np

# Criação do espaço de parâmetros genérico
eixo_x = np.linspace(0, 10, 100)
reta_equilibrio = eixo_x  # Representa p1/p2 = a1/a2

# Configuração da figura com proporções acadêmicas
plt.figure(figsize=(8, 6))

# Traço da linha de equilíbrio
plt.plot(eixo_x, reta_equilibrio, color='black', linewidth=2, label='Equilíbrio ($w_1 = w_2$)')

# Preenchimento das regiões topológicas de especialização
plt.fill_between(eixo_x, reta_equilibrio, 10, color='lightblue', alpha=0.4, 
                 label='Especialização no Bem 1 ($w_1 > w_2$)')
plt.fill_between(eixo_x, reta_equilibrio, 0, color='lightcoral', alpha=0.4, 
                 label='Especialização no Bem 2 ($w_2 > w_1$)')

# Remoção explícita de qualquer marcação numérica nos eixos
plt.xticks([])
plt.yticks([])

# Inserção dos rótulos algébricos nos eixos
plt.xlabel('Custo de Oportunidade ($a_1/a_2$)', fontsize=14)
plt.ylabel('Preço Relativo ($p_1/p_2$)', fontsize=14)

# Configuração de limites e estética limpa
plt.xlim(0, 10)
plt.ylim(0, 10)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Posição da legenda
plt.legend(loc='upper left', fontsize=11)

plt.show()
```

**2.5. Aplicação: preços relativos e salários**

Consideramos os seguintes dados hipotéticos:

* Unidades de trabalho: $L = 1000$ unidades
* Necessidade unitária de trabalho para o bem 1: $a_1 = 2$
* Necessidade unitária de trabalho para o bem 2: $a_2 = 4$
* Preço para o bem 1: $p_1 = 10$
* Preço para o bem 2: $p_2 = 20$

**Nota:** Usamos os mesmos dados de trabalho e necessidade unitária de trabalho do item 2.3

```python
# 2.5. Aplicação: preços relativos e salários
 
import matplotlib.pyplot as plt
import numpy as np

class MercadoTrabalhoRicardiano:
    def __init__(self, a1, a2, p1, p2):
        self.a1 = a1
        self.a2 = a2
        self.p1 = p1
        self.p2 = p2

    def calcular_salarios(self):
        w1 = self.p1 / self.a1
        w2 = self.p2 / self.a2
        return w1, w2

    def obter_parametros_grafico(self):
        preco_relativo = self.p1 / self.p2
        custo_oportunidade = self.a1 / self.a2
        return custo_oportunidade, preco_relativo

    def analisar_equilibrio(self):
        preco_relativo = self.p1 / self.p2
        custo_oportunidade = self.a1 / self.a2
        
        if preco_relativo > custo_oportunidade:
            return "Especialização total no Bem 1 (w1 > w2)"
        elif preco_relativo < custo_oportunidade:
            return "Especialização total no Bem 2 (w2 > w1)"
        else:
            return "Equilíbrio: Produção simultânea de ambos os bens (w1 = w2)"

# 1. setar a economia
modelo = MercadoTrabalhoRicardiano(a1=2, a2=4, p1=10, p2=20)
salario_1, salario_2 = modelo.calcular_salarios()
resultado_alocacao = modelo.analisar_equilibrio()
custo_op, preco_rel = modelo.obter_parametros_grafico()

# 2. resultados textuais
print(f"Salário no setor 1 (w1): {salario_1}")
print(f"Salário no setor 2 (w2): {salario_2}")
print(f"Custo de Oportunidade (a1/a2): {custo_op}")
print(f"Preço Relativo (p1/p2): {preco_rel}")
print(f"Resultado do mercado: {resultado_alocacao}")

# 3. gráfico
limite_eixo = max(custo_op, preco_rel) * 2.5
if limite_eixo == 0: limite_eixo = 2

eixo_x = np.linspace(0, limite_eixo, 100)
reta_equilibrio = eixo_x

plt.figure(figsize=(8, 6))

# Traço da linha de equilíbrio
plt.plot(eixo_x, reta_equilibrio, color='black', linewidth=2, label='Linha de Equilíbrio')

# Preenchimento das regiões de especialização
plt.fill_between(eixo_x, reta_equilibrio, limite_eixo, color='lightblue', alpha=0.4, 
                 label='Especialização Bem 1 ($w_1 > w_2$)')
plt.fill_between(eixo_x, reta_equilibrio, 0, color='lightcoral', alpha=0.4, 
                 label='Especialização Bem 2 ($w_2 > w_1$)')

# Inserção do ponto numérico exato do cenário atual
plt.scatter([custo_op], [preco_rel], color='darkred', s=120, zorder=5, 
            label=f'Cenário Atual ({custo_op:.1f}, {preco_rel:.1f})')

# Retorno das marcações numéricas para o caso prático
plt.xlabel('Custo de Oportunidade ($a_1/a_2$)', fontsize=12)
plt.ylabel('Preço Relativo ($p_1/p_2$)', fontsize=12)

# Configuração de limites e estética limpa
plt.xlim(0, limite_eixo)
plt.ylim(0, limite_eixo)
plt.grid(True, linestyle=':', alpha=0.7)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.legend(loc='upper left', fontsize=10)
plt.show()
```

#### **3. Economia de fator único: livre comércio**

A abertura da economia para o livre comércio envolve dois países, denominados local e estrangeiro.

* Mercado concorrencial;
* Ambos os países produzem os dois bens e utilizam como fator único o trabalho;
* Sem restrições à importação ou à exportação;
* Não há custos transacionais, barreiras tarifárias e não tarifárias, oriundos da distância ou heterogeneidade dos bens;
* A integração econômica indica a soma da produção e do consumo.

**Importante:** Partimos do pressuposto de que ambos os países possuem tamanhos similares como simplificação.



**3.1. Economia local e estrangeira**

Consideramos os seguintes dados hipotéticos:

Local (o mesmo do exemplo do item 2):

* Unidades de trabalho: $L = 1000$ unidades
* Necessidade unitária de trabalho para o bem 1: $a_1 = 2$
* Necessidade unitária de trabalho para o bem 2: $a_2 = 4$

Estrangeiro:

* Unidades de trabalho: $L^{*} = 1200$ unidades
* Necessidade unitária de trabalho para o bem 1: $a_1^{*}= 5$
* Necessidade unitária de trabalho para o bem 2: $a_2^{*} = 3$



```python
# 3.1. resultado base

import numpy as np
import matplotlib.pyplot as plt

# 1. Parâmetros Local
L = 1000
a1 = 2
a2 = 4

# 2. Parâmetros Estrangeiro
L_star = 1200
a1_star = 5
a2_star = 3

# 3. Capacidades Máximas de Produção (Especialização Absoluta)
Q1_max = L / a1
Q2_max = L / a2

Q1_star_max = L_star / a1_star
Q2_star_max = L_star / a2_star

# 4. Custos de Oportunidade (Preços Relativos em Autarquia)
custo_op_local = a1 / a2
custo_op_estrangeiro = a1_star / a2_star

# Exibição Numérica
print("*** Resultados Numéricos ***")
print(f"Custo de Oportunidade Local (Bem 1): {custo_op_local:.2f}")
print(f"Custo de Oportunidade Estrangeiro (Bem 1): {custo_op_estrangeiro:.2f}")
print(f"Produção Máxima Local (Bem 1: {Q1_max}, Bem 2: {Q2_max})")
print(f"Produção Máxima Estrangeiro (Bem 1: {Q1_star_max}, Bem 2: {Q2_star_max})")

# 5. Ponto de Especialização na Oferta Relativa Mundial
# A quantidade relativa é a oferta total do Bem 1 dividida pela oferta total do Bem 2
# quando ambos os países aplicam especialização absoluta.
Q_relativa_especializacao = Q1_max / Q2_star_max

# 6. Rotinas de Plotagem
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Gráfico 1: Fronteiras de Possibilidades de Produção (FPP)
ax1.plot([Q1_max, 0], [0, Q2_max], label='FPP Local', color='blue', linewidth=2)
ax1.plot([Q1_star_max, 0], [0, Q2_star_max], label='FPP Estrangeiro', color='red', linewidth=2)
ax1.set_xlabel('Quantidade do Bem 1')
ax1.set_ylabel('Quantidade do Bem 2')
ax1.set_title('Fronteiras de Possibilidades de Produção')
ax1.legend()
ax1.grid(True, linestyle=':', alpha=0.7)

# Gráfico 2: Oferta Relativa Mundial (RS)
# Vetores definem os degraus da curva de oferta relativa
precos = [0, custo_op_local, custo_op_local, custo_op_estrangeiro, custo_op_estrangeiro, 2.5]
quantidades = [0, 0, Q_relativa_especializacao, Q_relativa_especializacao, 3, 3]

ax2.step(quantidades, precos, where='post', color='green', linewidth=2, label='Oferta Relativa')
ax2.axhline(y=custo_op_local, color='blue', linestyle=':', alpha=0.5, label='Custo Oport. Local')
ax2.axhline(y=custo_op_estrangeiro, color='red', linestyle=':', alpha=0.5, label='Custo Oport. Estrangeiro')
ax2.axvline(x=Q_relativa_especializacao, color='purple', linestyle=':', alpha=0.5, label='Especialização Completa')

ax2.set_xlim(0, 2.5)
ax2.set_ylim(0, 2.5)
ax2.set_xlabel('Quantidade Relativa Mundial $(Q_1 + Q_1^*) / (Q_2 + Q_2^*)$')
ax2.set_ylabel('Preço Relativo Mundial $(p_1 / p_2)$')
ax2.set_title('Oferta Relativa Mundial')
ax2.legend()
ax2.grid(True, linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```

#### **4) Generalização do modelo para 'n' bens**

**Generalização:**

Podemos generalizar o modelo de vantagens comparativas para 'n' bens. Para isso, consideramos dois países (local e estrangeiro) que produzem 'n' bens. Assim, comparamos as unidades necessárias de trabalho do país local ($a_i$) e estrangeiro ($a_i^{*}$). Em seguida, ordenamos o custo de oportunidade:

$$\frac{a_1}{a_1^*} < \frac{a_2}{a_2^*} < \dots < \frac{a_n}{a_n^*} \tag{10}$$

**Obs.:** São argumentos que indicam o conflito de interesses e economia política do comércio internacional.

**Salário relativo:**

A ideia central para definir o padrão de comércio estabelece que a produção do bem ocorre no país de menor custo. Para definir o custo, adotamos como base o salário ($w_i$) e a necessidade unitária de trabalho ($a_i$).

$$\text{Custo de produção} = w_i a_i  \tag{11}$$

Agora, fazemos a comparação entre o custo no país local ($w_i a_i$) e estrangeiro ($w_i^{*} a_i^{*}$):

$$\begin{aligned}     w_i a_i &> w_i^{*} a_i^{*} \quad \text{(Custo local maior)} \\     w_i a_i &< w_i^{*} a_i^{*} \quad \text{(Custo estrangeiro maior)} \\     w_i a_i &= w_i^{*} a_i^{*} \quad \text{(Custos iguais)} \end{aligned} \tag{12}$$

Em seguida, simulamos o modelo para 100 bens ($n = 100$).

```python
# 4. Simulação do modelo para n bens

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Configuração da semente para reprodutibilidade
np.random.seed(42)

# Parâmetros da simulação
n_bens = 100
bens_ids = [f"Bem {i+1}" for i in range(n_bens)]
a_local = np.random.uniform(1, 50, n_bens)
a_estr = np.random.uniform(1, 50, n_bens)
w_local = 2.0
w_estr = 3.0

# Criação e ordenação do DataFrame
df = pd.DataFrame({
    "Bem": bens_ids,
    "a_i": a_local,
    "a_i*": a_estr
})
df["Custo Relativo"] = df["a_i"] / df["a_i*"]
df = df.sort_values(by="Custo Relativo").reset_index(drop=True)

# Definição do limite salarial
w_relativo = w_local / w_estr
salario_inverso = 1 / w_relativo

# Definição do produtor eficiente para a tabela
df["Produtor Eficiente"] = df["Custo Relativo"].apply(
    lambda x: "Local" if x < salario_inverso else ("Indiferente" if x == salario_inverso else "Estrangeiro")
)

# Exibição das tabelas (Extremos da cadeia produtiva)
print("Extremo de Maior Vantagem para o País Local (Top 5):")
print(df.head())
print("\n" + "-"*65 + "\n")
print("Extremo de Maior Vantagem para o País Estrangeiro (Top 5):")
print(df.tail())

# Definição de cores: Azul para Local, Vermelho para Estrangeiro
cores = ['#1f77b4' if x < salario_inverso else '#d62728' for x in df["Custo Relativo"]]

# Criação do gráfico
plt.figure(figsize=(12, 6))
plt.bar(df.index, df["Custo Relativo"], color=cores, width=1.0, edgecolor='black', linewidth=0.5)

# Linha de corte do salário relativo inverso
plt.axhline(y=salario_inverso, color='black', linestyle='--', linewidth=2, 
            label=f'Limite: Salário Inverso ($w^* / w = {salario_inverso:.2f}$)')

# Formatação visual
plt.title('Padrão de Comércio para 100 Bens (Modelo Ricardiano)', fontsize=14, pad=15)
plt.xlabel('Espectro de Bens (Ordenados por Custo Relativo)', fontsize=12)
plt.ylabel('Custo Relativo do Trabalho ($a_i / a_i^*$)', fontsize=12)
plt.ylim(0, df["Custo Relativo"].max() * 1.1)

# Legenda personalizada
patch_local = mpatches.Patch(color='#1f77b4', label='Produção Local')
patch_estr = mpatches.Patch(color='#d62728', label='Produção Estrangeira')
plt.legend(handles=[patch_local, patch_estr, plt.Line2D([0], [0], color='black', linestyle='--', linewidth=2, label=f'Limite ($w^* / w = {salario_inverso:.2f}$)')],
           loc='upper left', fontsize=11)

plt.grid(axis='y', linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```
